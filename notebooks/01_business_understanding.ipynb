# Hospital Readmission & Resource Allocation Analytics

### Optimizing Post-Discharge Resource Allocation for Patients with Diabetes

This project evaluates how 30-day readmission risk and healthcare utilization patterns can support the allocation of limited post-discharge resources for patients with diabetes.

Readmission prediction is treated as an input to decision-making rather than the final outcome. The analysis moves from identifying patterns associated with readmission to estimating patient risk, prioritizing patients, and evaluating resource-allocation strategies under different levels of care-management capacity.

The project follows the analytical framework:

**Healthcare Utilization → Readmission Risk → Patient Prioritization → Resource Allocation**


## 1.1 Business Problem

This project considers a resource-constrained post-discharge setting in which care-management capacity is insufficient to provide the same level of additional support to every patient.

The resulting decision problem is one of prioritization: which patients should receive additional post-discharge support first?

For patients with diabetes, the analysis evaluates whether 30-day readmission risk and healthcare utilization patterns can provide a structured basis for this prioritization. The objective is not simply to predict readmission, but to translate risk information into patient priority groups and evaluate how those groups could be supported under different levels of care-management capacity.

The primary decision addressed by the project is:

**Given limited post-discharge care-management capacity, which patients should be prioritized for additional support based on their 30-day readmission risk and healthcare utilization patterns?**


## 1.2 Stakeholders

The primary stakeholder is the Care Management or Population Health Team responsible for coordinating post-discharge support. The analysis is intended to help this team determine which patients should receive priority when the capacity for additional follow-up is limited.

Hospital Operations Leadership is a secondary stakeholder. The analysis provides an operational view of how readmission risk, healthcare utilization, patient prioritization, and available care-management capacity can be considered together when planning post-discharge support.


## 1.3 Business Questions

This project is guided by three questions:

1. **How can hospitals use 30-day readmission risk and healthcare utilization patterns to optimize the allocation of limited post-discharge resources for patients with diabetes?**

2. **Which patient characteristics and healthcare utilization patterns are associated with higher 30-day readmission risk?**

3. **How should patients be prioritized across different levels of post-discharge support when care-management capacity is limited?**


## 1.4 Analytical Approach

The analysis progresses through four connected stages.

Healthcare utilization analysis examines patient, encounter, clinical, treatment, and prior utilization characteristics in relation to 30-day readmission.

Readmission risk modeling evaluates whether the available information can be used to estimate relative differences in 30-day readmission risk. Predictive performance is assessed with emphasis on the model's ability to rank higher-risk patients rather than relying solely on a fixed classification threshold.

Patient prioritization translates predicted risk into an ordered patient population and evaluates how observed readmissions are distributed across progressively higher-risk groups.

Resource-allocation analysis examines how different levels of care-management capacity affect the number of patients who can receive additional support and the proportion of observed readmissions reached.

The final analytical outputs are translated into an interactive dashboard for exploring readmission patterns, patient priorities, and resource-allocation scenarios.


## 1.5 Project Scope

The project focuses on hospital encounters involving patients with diabetes, with readmission within 30 days of discharge as the primary outcome of interest.

### In Scope

- Patient, clinical, treatment, and encounter characteristics available in the dataset
- Prior inpatient, outpatient, and emergency healthcare utilization
- Analysis of factors associated with 30-day readmission
- Development and evaluation of predictive models for 30-day readmission risk
- Risk-based patient prioritization
- Evaluation of prioritization under different care-management capacity levels
- Scenario-based assessment of potential intervention impact
- Development of an interactive decision-support dashboard

### Out of Scope

- Determining the clinical cause of an individual readmission
- Determining whether an individual readmission was preventable
- Prescribing treatment or making individual clinical decisions
- Estimating causal effects of post-discharge interventions
- Estimating realized financial savings without validated cost and intervention-effectiveness data
- Deployment of the model within a clinical environment


## 1.6 Analytical Boundaries

The dataset is retrospective and observational. Relationships identified through the analysis are therefore interpreted as predictive associations rather than evidence of causal effects.

The unit of analysis is a hospital encounter, and individual patients may have multiple encounters. Patient identity is therefore considered when separating data for model development and evaluation to prevent the same patient from appearing across training and test populations.

The analysis is limited to information available in the dataset. Healthcare utilization variables represent observed prior utilization and are not treated as complete measures of healthcare resource consumption, patient need, or cost.

Predicted probabilities represent relative model-estimated readmission risk rather than certainty that an individual patient will or will not be readmitted. Patient prioritization is therefore evaluated as a mechanism for directing additional support rather than determining access to standard care.

Care-management capacity and intervention-effectiveness scenarios are analytical simulations. They do not represent the staffing levels, intervention effectiveness, costs, or operating conditions of a specific hospital.

The resulting framework is intended as a decision-support analysis. Operational use would require validation with contemporary local data, prospective assessment, and appropriate clinical and organizational review.
