# Hospital Readmission & Resource Allocation Analytics

### Optimizing Post-Discharge Resource Allocation for Patients with Diabetes

This project examines how 30-day hospital readmission risk and healthcare utilization patterns can be used to support the allocation of limited post-discharge resources for patients with diabetes.

Rather than treating readmission prediction as the final outcome, the project uses predictive analytics as an input into a broader decision-support framework. The goal is to identify patients at elevated risk of 30-day readmission, understand their healthcare utilization patterns, and translate these insights into patient prioritization strategies under limited care-management capacity.

## 1.1 Project Background

Hospital readmissions are an important challenge for healthcare organizations because they can reflect complex patient needs, continuity-of-care challenges, and repeated demand on hospital services.

For patients with diabetes, post-discharge care can be particularly important because diabetes management may involve multiple medications, comorbidities, ongoing monitoring, and interactions with different parts of the healthcare system.

At the same time, hospitals operate with limited post-discharge resources, including care-management capacity, follow-up services, and clinical staff time. Providing the same level of support to every discharged patient may therefore be neither practical nor an optimal use of available resources.

A data-driven approach can help hospitals understand which patients demonstrate greater readmission risk and how previous healthcare utilization relates to that risk. These insights can then support more informed prioritization of limited post-discharge resources.

## 1.2 Business Problem

Hospitals must make post-discharge support decisions under limited care-management capacity. Not every patient requires the same level of follow-up, yet identifying which patients should receive priority can be challenging.

For patients with diabetes, differences in prior hospital utilization, clinical complexity, treatment patterns, and encounter characteristics may signal varying levels of 30-day readmission risk.

The business challenge is therefore not simply to predict which patients may be readmitted, but to determine how readmission risk and healthcare utilization information can support patient prioritization when post-discharge resources are constrained.

This project addresses that challenge by developing an analytical framework that connects:

**Healthcare Utilization → Readmission Risk → Patient Prioritization → Resource Allocation**

## 1.3 Stakeholders

### Primary Stakeholder
**Care Management / Population Health Team**

The primary stakeholder is the team responsible for coordinating post-discharge support and managing patients who may require additional follow-up after hospitalization. Their key decision is how to prioritize patients when care-management capacity is limited.

### Secondary Stakeholder
**Hospital Operations Leadership**

Hospital operations leaders have an interest in understanding patterns of readmission and healthcare utilization, as well as how available post-discharge resources can be aligned with patient needs.

### Decision to Support

The analysis is designed to support the following operational decision:

> **Given limited post-discharge care-management capacity, which patients should be prioritized for additional support based on their 30-day readmission risk and healthcare utilization patterns?**

## 1.4 Business Questions

This project is guided by the following questions:

1. **How can hospitals use 30-day readmission risk and healthcare utilization patterns to optimize the allocation of limited post-discharge resources for patients with diabetes?**

2. **Which patient characteristics and healthcare utilization patterns are associated with higher 30-day readmission risk?**

3. **How should patients be prioritized across different levels of post-discharge support when care-management capacity is limited?**

## 1.5 Analytical Objectives

To address the business questions, the project will:

1. Examine the distribution and patterns of 30-day hospital readmissions among patients with diabetes.

2. Analyze patient characteristics, encounter characteristics, and healthcare utilization patterns associated with 30-day readmission.

3. Develop and evaluate predictive models that estimate the probability of 30-day readmission.

4. Combine predicted readmission risk with healthcare utilization patterns to develop meaningful patient priority segments.

5. Evaluate how patients could be prioritized under different levels of post-discharge care-management capacity.

6. Develop an interactive decision-support dashboard that communicates readmission risk, utilization patterns, patient priorities, and resource-allocation scenarios.
